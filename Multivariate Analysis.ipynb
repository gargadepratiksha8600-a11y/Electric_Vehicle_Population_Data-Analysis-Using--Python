```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Assume your DataFrame is named df

# --- Remove outliers using IQR ---
def remove_outliers(df, column):
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    return df[(df[column] >= lower) & (df[column] <= upper)]

# Apply for both columns
df_clean = remove_outliers(df, 'Base MSRP')
df_clean = remove_outliers(df_clean, 'Electric Range')

# --- Plot after removing outliers ---
plt.figure(figsize=(8,5))
sns.scatterplot(data=df_clean, x='Base MSRP', y='Electric Range',
                hue='Electric Vehicle Type', alpha=0.7)
plt.title('Base MSRP vs Electric Range (Outliers Removed)')
plt.xlabel('Base MSRP')
plt.ylabel('Electric Range')
plt.legend(title='Electric Vehicle Type')
plt.show()
```
