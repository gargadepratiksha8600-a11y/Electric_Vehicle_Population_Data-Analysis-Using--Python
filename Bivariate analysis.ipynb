```python
import pandas as pd
import seaborn as sns
```

```python
df = pd.read_csv(r"C:\Users\dell\Downloads\archive\Electric_Vehicle_Population_Data.csv")
```

```python
df
```

```python
#1.Count of electric vehicle types by County.
result = df.groupby(["County", "Electric Vehicle Type"]).size().reset_index(name="Count")

print(result)
```

```python
#2.Count of electric vehicle types by City.

result = df.groupby(["City", "Electric Vehicle Type"]).size().reset_index(name="Count")

print(result)
```

```python
#3 Count of electric vehicle types by Make
count_by_make = df.groupby(["Make", "Electric Vehicle Type"]).size().reset_index(name="Count")
print(count_by_make)
```

```python
#4 Average Electric Range by Make
avg_range_by_make = df.groupby("Make")["Electric Range"].mean().reset_index()
print(avg_range_by_make)
```

```python
#5 Average Base MSRP by Make
avg_msrp_by_make = df.groupby("Make")["Base MSRP"].mean()
print(avg_msrp_by_make)
```

```python
#6. Average Electric Range by Electric Vehicle Type
avg_range_by_type = df.groupby("Electric Vehicle Type")["Electric Range"].mean()
print(avg_range_by_type)
```

```python
#7. Average Base MSRP by Electric Vehicle Type
avg_msrp_by_type = df.groupby("Electric Vehicle Type")["Base MSRP"].mean()
print(avg_msrp_by_type)
```

```python
#8. Distribution of Model Year by Make
model_year_distribution = df.groupby(["Make", "Model Year"]).size().reset_index(name="Count")
print(model_year_distribution)
```

```python
#9 Average Electric Range by Model Year
avg_range_by_year = df.groupby("Model Year")["Electric Range"].mean()
print(avg_range_by_year)
```

```python
#10. Average Base MSRP by Model Year
avg_msrp_by_year = df.groupby("Model Year")["Base MSRP"].mean()
print(avg_msrp_by_year)
```

```python
#11.Top 5 counties by total Electric Range sum
top5_counties_range = (
    df.groupby("County")["Electric Range"]
      .sum()
      .reset_index()
      .sort_values(by="Electric Range", ascending=False)
      .head(5)
)
print(top5_counties_range)
```

```python
#12. Count of vehicles by CAFV Eligibility and Electric Vehicle Type
count_cafv_evtype = (
    df.groupby(["Clean Alternative Fuel Vehicle (CAFV) Eligibility", "Electric Vehicle Type"])
      .size()
      .reset_index(name="Count")
)
print(count_cafv_evtype)
```

```python
# 13. Count of vehicles by CAFV Eligibility and County
count_cafv_county = (
    df.groupby(["Clean Alternative Fuel Vehicle (CAFV) Eligibility", "County"])
      .size()
      .reset_index(name="Count")
)
print(count_cafv_county)
```

```python
#14. Number of unique Models by Make
unique_models_by_make = df.groupby("Make")["Model"].nunique().reset_index(name="Unique Models")
print(unique_models_by_make)
```

```python
#15# 2. Count of vehicles by State and Electric Vehicle Type
count_state_evtype = df.groupby(["State", "Electric Vehicle Type"]).size().reset_index(name="Count")
print(count_state_evtype)
```

```python
#16. Trend of Electric Range change over Model Year for top 5 makes
top5_makes = df["Make"].value_counts().head(5).index
trend_range_top5 = (
    df[df["Make"].isin(top5_makes)]
    .groupby(["Model Year", "Make"])["Electric Range"]
    .mean()
    .reset_index()
    .sort_values(["Make", "Model Year"])
)
print(top5_makes)
```

```python
#17. Top 5 most expensive Base MSRP models per Make
top5_expensive_by_make = (
    df.groupby("Make", group_keys=False)
      .apply(lambda x: x.nlargest(5, "Base MSRP"))
      [["Make", "Model", "Base MSRP"]]
      .reset_index(drop=True)
)
print(top5_expensive_by_make)
```

```python
#18. Count of Electric Vehicle Types by County (Top 10)
top_counties = df["County"].value_counts().head(10).index
sns.countplot(data=df[df["County"].isin(top_counties)],
x="County", hue="Electric Vehicle Type")
```

```python
#19 Average Electric Range by Make (Top 10)
top_makes = df["Make"].value_counts().head(10).index
sns.barplot(data=df[df["Make"].isin(top_makes)], x="Make", y="Electric Range")
```

```python
#20. Average Base MSRP by Electric Vehicle Type
sns.barplot(data=df, x="Electric Vehicle Type", y="Base MSRP")
```

```python
#21. Average Electric Range by Model Year
sns.lineplot(data=df.groupby("Model Year")["Electric Range"].mean().reset_index(),
             x="Model Year", y="Electric Range", marker="o")
```

```python
#22. Trend of Electric Range by Model Year for Top 5 Makes
top5 = df["Make"].value_counts().head(5).index
sns.lineplot(data=df[df["Make"].isin(top5)].groupby(["Model Year","Make"])
             ["Electric Range"].mean().reset_index(),
             x="Model Year", y="Electric Range", hue="Make", marker="o")
```

```python
import matplotlib.pyplot as plt
```

```python
#23. Count each Electric Vehicle Type
ev_counts = df["Electric Vehicle Type"].value_counts()

# Plot pie chart
plt.pie(ev_counts, labels=ev_counts.index, autopct="%1.1f%%", startangle=140)
plt.title("Share of Each Electric Vehicle Type")
plt.show()
```

```python
#24 Get top 5 most common model years
top_years = df["Model Year"].value_counts().head(5)
colors = sns.color_palette("pastel", len(top_years))
plt.pie(top_years, labels=top_years.index, autopct="%1.1f%%", startangle=140, colors=colors)
plt.title("Share of Model Year Distribution (Top 5 Years)")
plt.show()
```

```python
# 25. Count of vehicles by Postal Code and County
count_by_postal_county = df.groupby(["Postal Code", "County"]).size().reset_index(name="Count")
print(count_by_postal_county)
```

```python
#26. std Electric Range by Postal Code
avg_range_postal = df.groupby("Postal Code")["Electric Range"].std()
print(avg_range_postal)
```

```python
#27 Count of vehicles by Electric Utility and County
count_by_utility_county = df.groupby(["Electric Utility", "County"]).size().reset_index(name="Count")
print(count_by_utility_county)
```

```python
#28. Max and Min Electric Range for each Make
range_by_make = df.groupby("Make")["Electric Range"].agg(Max_Range="max", Min_Range="min").reset_index()
print(range_by_make)
```

```python
#29. Max and Min Base MSRP for each Electric Vehicle Type
msrp_by_type = df.groupby("Electric Vehicle Type")["Base MSRP"].agg(Max_MSRP="max", Min_MSRP="min").reset_index()
print(msrp_by_type)
```

```python
#30. Max and Min Model Year for each County
year_by_county = df.groupby("County")["Model Year"].agg(Max_Year="max", Min_Year="min").reset_index()
print(year_by_county)
```

```python
#2.Count of electric vehicle types by top 5 City.

top_2_by_type = (
    df.groupby(["Electric Vehicle Type", "City"])
      .size()
      .reset_index(name="Count")
      .sort_values(["Electric Vehicle Type", "Count"], ascending=[True, False])
      .groupby("Electric Vehicle Type")
      .head(2)
)

print(top_2_by_type)
```
